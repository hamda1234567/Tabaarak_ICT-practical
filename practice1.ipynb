```python
#copy
new_df=df.copy()
```

```python
#fiilna in categorical column gender
new_df['gender'].mode()
```

```python
new_df['gender']=new_df['gender'].fillna(new_df['gender'].mode()[0])
new_df
```

```python
# fillna in numrical  coulumn age
new_df['age'].median()
```

```python
new_df['age']=new_df['age'].fillna(new_df['age'].median()).astype (int)
```

```python
# fillna in categorical coulumn city
new_df['city'].mode()
```

```python
new_df['city']=new_df['city'].fillna(new_df['city'].mode()[0])
new_df
```

```python
new_df['attendance_percent'].median()
```

```python
new_df['attendance_percent']=new_df['attendance_percent'].fillna(new_df
 ['attendance_percent'].median()).astype (int)
```

```python
new_df['exam_score'].median()
```

```python
new_df['exam_score']=new_df['exam_score'].fillna(new_df
 ['exam_score'].median()).astype (int)
```

```python
new_df.isnull().sum()
```

```python
# Count duplicate rows
duplicates = new_df.duplicated().sum()
print("Number of duplicates:", duplicates)
```

```python
#drop dulicates
new_df=new_df.drop_duplicates()
```

```python
new_df.head()
```

```python
new_df['gender'] = new_df['gender'].astype(str).str.strip().str.lower()

new_df['gender'] = new_df['gender'].replace({
    'm': 'Male',
    'male': 'Male',
    'f': 'Female',
    'female': 'Female'
})

print(new_df['gender'].value_counts(dropna=False))
```

```python
new_df['internet_access'] = (
    new_df['internet_access']
    .astype(str)
    .str.strip()
    .str.lower()
)

new_df['internet_access'] = new_df['internet_access'].replace({
    'yes': 'Yes',
    'no': 'No'
})

print(new_df['internet_access'].value_counts(dropna=False))
```

```python
new_df['parent_support'] = (
    new_df['parent_support']
    .astype(str)
    .str.strip()
    .str.lower()
)

new_df['parent_support'] = new_df['parent_support'].replace({
    'yes': 'Yes',
    'no': 'No'
})

print(new_df['parent_support'].value_counts(dropna=False))
```

```python
import matplotlib.pyplot as plt
gender_counts = new_df['gender'].value_counts()
plt.bar(gender_counts.index, gender_counts.values)
plt.xlabel('Gender')
plt.ylabel('Number of Students')
plt.title('Students by Gender')

plt.show()
```

```python
new_df.to_csv("  pracical_cleaned.csv ",index=False)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler,OneHotEncoder
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
```

```python
numeric_features = [
    "age",
    "study_hours_per_day",
    "attendance_percent",
    "absences",
    "exam_score"
]

categorical_features = [
    "gender",
    "school",
    "city",
    "internet_access",
    "parent_support"
]

preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numeric_features),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features)
    ],
    remainder='drop'
)
```

```python

#seperate feature (x) and target (y)
x=new_df.drop("grade",axis=1)
y=new_df["grade"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(x, y,
                                                    test_size=0.2, random_state=42)
```

```python
# Create a pipeline for Logistic Regression
log_reg_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', LogisticRegression(random_state=42))
])

# Train the Logistic Regression model
log_reg_pipeline.fit(X_train, y_train)
```

```python
# Make predictions with Logistic Regression
y_pred_log_reg = log_reg_pipeline.predict(X_test)

# Evaluate Logistic Regression model
print("Logistic Regression Model Evaluation:")
print(f"Accuracy: {accuracy_score(y_test, y_pred_log_reg):.4f}")
print(f"Precision: {precision_score(y_test, y_pred_log_reg, average='weighted'):.4f}")
print(f"Recall: {recall_score(y_test, y_pred_log_reg, average='weighted'):.4f}")
print(f"F1-Score: {f1_score(y_test, y_pred_log_reg, average='weighted'):.4f}")
```

```python
# Create a pipeline for Decision Tree Classifier
dec_tree_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', DecisionTreeClassifier(random_state=42))
])

# Train the Decision Tree model
dec_tree_pipeline.fit(X_train, y_train)
```

```python
# Make predictions with Decision Tree
y_pred_dec_tree = dec_tree_pipeline.predict(X_test)

# Evaluate Decision Tree model
print("\nDecision Tree Classifier Model Evaluation:")
print(f"Accuracy: {accuracy_score(y_test, y_pred_dec_tree):.4f}")
print(f"Precision: {precision_score(y_test, y_pred_dec_tree, average='weighted'):.4f}")
print(f"Recall: {recall_score(y_test, y_pred_dec_tree, average='weighted'):.4f}")
print(f"F1-Score: {f1_score(y_test, y_pred_dec_tree, average='weighted'):.4f}")
```
