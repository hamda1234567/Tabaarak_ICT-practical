**day 3 model training and evaluation**

```python

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler,OneHotEncoder
from sklearn.tree import DecisionTreeClassifier
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score,precision_score ,recall_score,f1_score
```

```python
numeric_features = [
    "studytime",
    "failures",
    "famrel",
    "health",
    "absences"
]

categorical_features = [
    "schoolsup",
    "famsup",
    "activities",
    "higher",
    "internet"
]
```

```python
#seperate feature (x) and target (y)
x=model_data.drop("at_risk",axis=1)
y=model_data["at_risk"]
```

```python
from sklearn import preprocessing
#preprocessing
preprocessr=ColumnTransformer(
    transformers=[
        ('cat',OneHotEncoder(),categorical_features),
        ('num',StandardScaler(),numeric_features)
                  ]
)
```

```python
from sklearn import pipeline
from sklearn.tree import DecisionTreeClassifier
#build pipeline
Pipeline=  pipeline.Pipeline(steps=[
    ('preprocessing',preprocessr),
    ('class', DecisionTreeClassifier())
])
```

```python
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
Pipeline.fit(X_train, y_train)
```

```python
y_pred = Pipeline.predict(X_test)
```

```python
y_pred
```

model **evaluation **bold text**

```python
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy Score: {accuracy}")

precision = precision_score(y_test, y_pred)
print(f"Precision Score: {precision}")

from sklearn.metrics import recall_score
recall = recall_score(y_test, y_pred)
print(f"Recall Score: {recall}")
f1 = f1_score(y_test, y_pred)
print(f"F1 Score: {f1}")
```

## Logistic Regression Model

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

# Build Logistic Regression pipeline
logistic_pipeline = Pipeline(steps=[
    ('preprocessing', preprocessr),
    ('classifier', LogisticRegression(random_state=42))
])
```

```python
# Fit the Logistic Regression pipeline
logistic_pipeline.fit(X_train, y_train)
```

```python
# Make predictions with Logistic Regression
y_pred_lr = logistic_pipeline.predict(X_test)
```

```python
# Evaluate Logistic Regression Model
accuracy_lr = accuracy_score(y_test, y_pred_lr)
precision_lr = precision_score(y_test, y_pred_lr)
recall_lr = recall_score(y_test, y_pred_lr)
f1_lr = f1_score(y_test, y_pred_lr)

print(f"Logistic Regression Accuracy Score: {accuracy_lr}")
print(f"Logistic Regression Precision Score: {precision_lr}")
print(f"Logistic Regression Recall Score: {recall_lr}")
print(f"Logistic Regression F1 Score: {f1_lr}")
```
